## 1. リングのパラメータ (`display/KEKLS_ring_params.sad`)

RINGC (標準セル 20 個) と RING (入射セルつき) の周長・チューン・エミッタンス・自然クロマティシティとクロマティシティを印字します (6 極は一時的に切って戻します)。

```python
%%sad
! 元の例のスクリプトと同じフラグ (FFS の前に置く)
ON RAD FLUC COD RFSW RADCOD;
FFS;

Chrom[dd_]:=(
  DP0 = dd;   FFS["cell calc"];  nxp = Twiss["NX","$$$"]/2/Pi;  nyp = Twiss["NY","$$$"]/2/Pi;
  DP0 = -dd;  FFS["cell calc"];  nxm = Twiss["NX","$$$"]/2/Pi;  nym = Twiss["NY","$$$"]/2/Pi;
  DP0 = 0;    FFS["cell calc"];
  {(nxp-nxm)/2/dd, (nyp-nym)/2/dd}
);

names = {"RINGC", "RING"};
Do[
  FFS["USE " // names[k]];
  Print["===== ", names[k]];
  FFS["cell calc emit",6];
  Print["circumference [m]  = ", LINE["S","$$$"]];
  Print["tune               = ", Twiss["NX","$$$"]/2/Pi, " ", Twiss["NY","$$$"]/2/Pi];
  Print["EMITX [pm rad]     = ", EMITX*1E12];
  Print["SIGE               = ", SIGE];
  Print["SIGZ [mm]          = ", SIGZ*1000];
  xi = Chrom[0.0005];
  Print["xi (with sextupoles)    = ", xi[1], " ", xi[2]];
  sdh0 = LINE["K2","SDH"];  sfh0 = LINE["K2","SFH"];
  LINE["K2","SDH"] = 0;  LINE["K2","SFH"] = 0;
  FFS["cell calc"];
  xi = Chrom[0.0005];
  Print["xi (natural, K2 = 0)    = ", xi[1], " ", xi[2]];
  LINE["K2","SDH"] = sdh0;  LINE["K2","SFH"] = sfh0;
  FFS["cell calc"];
,{k,1,2}];
```

## 2. β 関数と分散関数を描く

RING の結果を `sad.value()` で取り出して、Python で描きます。

```python
import matplotlib.pyplot as plt
s  = sad.value('Twiss["S","*"]')
bx = sad.value('Twiss["BX","*"]'); by = sad.value('Twiss["BY","*"]'); ex = sad.value('Twiss["EX","*"]')
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(s, bx, label='betax [m]'); ax.plot(s, by, label='betay [m]'); ax.plot(s, [100 * e for e in ex], label='etax x 100 [m]')
ax.set_xlabel('s [m]'); ax.legend(); fig.tight_layout(); plt.show()
print('要素の数', len(s), ' 周長 [m]', s[-1])
```

## 3. クロマティシティ (`display/KEKLS_chromaticity.sad`)

δ を ±3% まで振ったチューンを、6 極あり・なしで求めます (6 極の K2 は戻します)。

```python
%%sad --continue
USE RINGC;
cell calc emit;
dd = 0.005;
nd = 12;
Print["--- with the design sextupoles (SDH, SFH)   columns: delta nux nuy"];
Do[
  DP0 = (i - nd/2) * dd;
  FFS["cell calc"];
  Print["T1 ", DP0, " ", Twiss["NX","$$$"]/2/Pi, " ", Twiss["NY","$$$"]/2/Pi];
,{i,0,nd}];
DP0 = 0;
FFS["cell calc"];
sdh0 = LINE["K2","SDH"];  sfh0 = LINE["K2","SFH"];
LINE["K2","SDH"] = 0;  LINE["K2","SFH"] = 0;
Print["--- without sextupoles (K2 = 0)   columns: delta nux nuy"];
Do[
  DP0 = (i - nd/2) * dd;
  FFS["cell calc"];
  Print["T0 ", DP0, " ", Twiss["NX","$$$"]/2/Pi, " ", Twiss["NY","$$$"]/2/Pi];
,{i,0,nd}];
DP0 = 0;
LINE["K2","SDH"] = sdh0;  LINE["K2","SFH"] = sfh0;
FFS["cell calc"];
```
